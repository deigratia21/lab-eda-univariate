```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('amz_uk_price_prediction_dataset.csv')

# Lihat 5 baris pertama
df.head()
```

```python
# Cek info kolom dan data types
df.info()

# Cek nilai yang hilang (missing values)
df.isnull().sum()
```

```python
# Frequency table of product categories
category_counts = df['category'].value_counts()
category_counts.head(10)
# Top 5 product categories
top5_categories = category_counts.head(5)
top5_categories
# Bar chart for all categories (you can limit to top N if it's too crowded)
plt.figure(figsize=(12,6))
sns.barplot(x=top5_categories.index, y=top5_categories.values)
plt.title('Top 5 Most Listed Product Categories')
plt.ylabel('Number of Listings')
plt.xlabel('Product Category')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
# Pie chart for top 5 categories
plt.figure(figsize=(8,8))
plt.pie(top5_categories.values, labels=top5_categories.index, autopct='%1.1f%%', startangle=140)
plt.title('Top 5 Categories Proportion')
plt.show()
```

```python
# Mean
mean_price = df['price'].mean()

# Median
median_price = df['price'].median()

# Mode (ambil nilai pertama dari mode yang bisa lebih dari satu)
mode_price = df['price'].mode()[0]

mean_price, median_price, mode_price
# Variance
price_variance = df['price'].var()

# Standard Deviation
price_std = df['price'].std()

# Range
price_range = df['price'].max() - df['price'].min()

# IQR (Interquartile Range)
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
price_iqr = Q3 - Q1

price_variance, price_std, price_range, price_iqr
plt.figure(figsize=(10,6))
sns.histplot(df[df['price'] < 200]['price'], bins=100, kde=True)
plt.title('Price Distribution Histogram (Prices < 200)')
plt.xlabel('Price')
plt.ylabel('Count')
plt.show()
plt.figure(figsize=(10,2))
sns.boxplot(x=df['price'])
plt.title('Boxplot of Product Prices')
plt.show()
```

```python
# Mean
mean_rating = df['stars'].mean()

# Median
median_rating = df['stars'].median()

# Mode
mode_rating = df['stars'].mode()[0]

mean_rating, median_rating, mode_rating
# Variance
rating_variance = df['stars'].var()

# Standard Deviation
rating_std = df['stars'].std()

# IQR
Q1_rating = df['stars'].quantile(0.25)
Q3_rating = df['stars'].quantile(0.75)
rating_iqr = Q3_rating - Q1_rating

rating_variance, rating_std, rating_iqr
# Skewness and Kurtosis
rating_skew = df['stars'].skew()
rating_kurt = df['stars'].kurt()

rating_skew, rating_kurt
plt.figure(figsize=(10,6))
sns.histplot(df['stars'], bins=20, kde=True)
plt.title('Distribution of Product Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```
